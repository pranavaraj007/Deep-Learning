```python

import matplotlib.pyplot as plt
import numpy as np

# Simulated training results (demonstration only)
epochs = [1, 2, 3, 4, 5]

results = {
    "VGG16": {
        "train": [48.5, 61.2, 70.4, 77.8, 83.1],
        "val":   [45.0, 56.0, 64.0, 69.0, 74.0],
        "loss":  [1.42, 1.10, 0.82, 0.63, 0.48]
    },
    "ResNet18": {
        "train": [52.0, 66.5, 75.2, 82.1, 87.4],
        "val":   [49.0, 61.0, 69.0, 75.0, 80.0],
        "loss":  [1.30, 0.95, 0.68, 0.49, 0.35]
    },
    "GoogLeNet": {
        "train": [50.1, 64.0, 73.5, 80.0, 85.2],
        "val":   [47.0, 59.0, 67.0, 72.0, 77.0],
        "loss":  [1.35, 1.00, 0.74, 0.55, 0.40]
    }
}

print("Using device: cpu")
print("Dataset: CIFAR-10")
print("Training images: 400")
print("Validation images: 100")

for name, values in results.items():
    print(f"\nTraining {name}...")
    for i in range(5):
        print(
            f"{name} Epoch [{i+1}/5] "
            f"- Loss: {values['loss'][i]:.4f} "
            f"- Train Acc: {values['train'][i]:.2f}% "
            f"- Val Acc: {values['val'][i]:.2f}%"
        )

# Validation accuracy graph
plt.figure(figsize=(9, 5))
for name, values in results.items():
    plt.plot(epochs, values["val"], marker='o', label=name)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.xticks(epochs)
plt.ylim(0, 100)
plt.legend()
plt.grid(True)
plt.show()

# Training accuracy graph
plt.figure(figsize=(9, 5))
for name, values in results.items():
    plt.plot(epochs, values["train"], marker='o', label=name)

plt.title("Training Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.xticks(epochs)
plt.ylim(0, 100)
plt.legend()
plt.grid(True)
plt.show()

# Final validation accuracy
print("\nFinal Validation Accuracy (Simulated):")
for name, values in results.items():
    print(f"{name}: {values['val'][-1]:.2f}%")

# Mock custom image predictions
print("\nPrediction results for: download.jpeg")
print("VGG16      => cat")
print("RESNET18   => cat")
print("GOOGLENET  => dog")

print("\nExp 5 demonstration completed.")
```
